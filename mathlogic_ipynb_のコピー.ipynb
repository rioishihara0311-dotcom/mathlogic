## (準備) 自然数の有限列のコーディング

ここでは，0 以上の整数を自然数とします。自然数の有限列全体を $\mathbb N^*$ と書き，この講義を通して $\mathbb N^*$ のコードとして用いる全単射

$$
\texttt{listcode}:\mathbb N^*\to\mathbb N
$$

を定義します。

まず，自然数 $a$ を $0,1$ の列で表現する次の方法を考えます。$a+1$ の2進表現の桁数から1を引いた数を $k$ とします。$a+1$ の2進表現から先頭の $1$ を除いた $k$ 桁の文字列，その後に $1$，さらに $k$ 個の $0$ を並べて得られる $2k+1$ 桁の $0,1$ 列を $c(a)$ とします。

たとえば，

$$
\begin{array}{c|c|c}
a & a+1 & c(a)\\ \hline
0 & 1   & 1\\
1 & 10  & 010\\
2 & 11  & 110\\
3 & 100 & 00100\\
4 & 101 & 01100\\
5 & 110 & 10100
\end{array}
$$

です。

空列のコードを

$$
\texttt{listcode}([])=0
$$

とします。また，

$$
\texttt{listcode}([a_0,\ldots,a_{n-1}])
$$

を，文字列

$$
c(a_{n-1})\cdots c(a_1)c(a_0)
$$

が2進表現として表す自然数とします。通常の2進表現と同様，先頭の $0$ は数の値に影響しません。

たとえば，

$$
\texttt{listcode}([0,0,0])=(111)_2=7,
$$

$$
\texttt{listcode}([1,2,3])=(00100\,110\,010)_2=(100110010)_2=306.
$$

このコードは右端から一意に復号できます。末尾の $0$ の個数を $k$ とすると，その直前の $1$ が区切りです。さらにその左の $k$ 桁を $r$ と読めば，先頭要素は

$$
a=2^k+r-1
$$

です。それより左に残った部分が残りの列のコードになります。この操作を繰り返すと最後に $0$ に到達します。したがって，これは $\mathbb N^*$ と $\mathbb N$ の全単射になっています。

````python
def cons(a, x):
    # a を，コード x が表す列の先頭につける
    n = a + 1
    p = 1
    while p + p <= n:
        p = p + p
    r = n - p
    return p * (2 * r + 1 + 2 * p * x)


def head(x):
    # コード x が表す列の先頭要素。空列では 0 を返す
    if x == 0:
        return 0

    p = 1
    while x % 2 == 0:
        x = x // 2
        p = p + p

    y = (x - 1) // 2
    r = y % p
    return p + r - 1


def tail(x):
    # コード x が表す列から先頭要素を除いた列のコード
    if x == 0:
        return 0

    p = 1
    while x % 2 == 0:
        x = x // 2
        p = p + p

    y = (x - 1) // 2
    return y // p


def listcode(u):
    # Python の有限リスト u を自然数にコード化する
    x = 0
    i = len(u)
    while i > 0:
        i = i - 1
        x = cons(u[i], x)
    return x


def listdecode(x):
    # 確認用：自然数コード x を Python のリストに戻す
    u = []
    while x > 0:
        u.append(head(x))
        x = tail(x)
    return u
````

````python
print(listcode([]))
print(listcode([0,0,0]))
print(listcode([1,2,3]))
print(listdecode(listcode([1,2,3])))

u = [3222, 26, 24886, 106, 1]
x = listcode(u)
print(x)
print(listdecode(x))
````

## (準備) コード化されたリストへのアクセス

`listentry(x, k)` は，数 `x` でコード化されたリストの `k` 番目の要素を返します。リストの長さ以上の位置にアクセスした場合は `0` を返すことにします。

`listlength(x)` は，コード化されたリストの長さを返します。

`listreplace(x, k, a)` は，`x` でコード化されたリストの `k` 番目の要素を `a` に置き換えたリストのコードを返します。`k` がリストの長さ以上なら，途中を `0` で補います。

````python
def listentry(x, k):
    while k > 0:
        x = tail(x)
        k = k - 1
    return head(x)


def listlength(x):
    n = 0
    while x > 0:
        x = tail(x)
        n = n + 1
    return n


def listreplace(x, k, a):
    kk = k
    u = 0

    # 最初の k 要素を逆順に u に退避する
    while k > 0:
        u = cons(head(x), u)
        x = tail(x)
        k = k - 1

    # k 番目を a に置き換える
    x = cons(a, tail(x))

    # 退避した prefix を元に戻す
    while kk > 0:
        x = cons(head(u), x)
        u = tail(u)
        kk = kk - 1

    return x


z = listcode([1,2,3,10,5])
print(listentry(z, 3))
u = listreplace(z, 4, 8)
print(listdecode(u))

z = listcode([1,2,3,4,5])
print(listentry(z, 10))
u = listreplace(z, 9, 8)
print(listdecode(u))
````

# ユニバーサル Abacus マシン(その1)

授業プリントと同じプログラムを python で書いたものです。
プリントでは p, d として自然数を与えているが，ここでは python のリストで与えています。よって，リストの長さより先にアクセスできないので，データ d としては，プログラム中に現れる箱の番号を全て含むように，十分長く 0 を後ろにつけたものを与える必要があります。


````python
def univ(p, d):
    pc = 0
    while(p[pc][0] > 0):
        op = p[pc][0]
        box = p[pc][1]
        next = p[pc][2]
        if(op == 1):
            d[box] += 1
            pc = next
        else:
            enext = p[pc][3]
            if(d[box] > 0):
                d[box] -= 1
                pc = next
            else:
                pc = enext
    return(d[0])
````

# ユニバーサル Abacus マシン(その2)

listentry, listreplace を用いて，上の univ のプログラムにおいて，プログラム p とデータ d を，リストではなく自然数のコードを引数とするように書きかえたもの。

````python
def univm(p, d):
    ''' p, d は，リストを数で表した数が与えられる '''

    pc = 0
    while(listentry(listentry(p, pc), 0) > 0):
        print(pc, listdecode(d))  # デバッグのためのプリント
        op = listentry(listentry(p, pc), 0)
        box = listentry(listentry(p, pc), 1)
        next = listentry(listentry(p, pc), 2)
        if(op == 1):
            d = listreplace(d, box, listentry(d, box)+1)
            pc = next
        else:
            enext = listentry(listentry(p, pc), 3)
            if(listentry(d,box) > 0):
                if(listentry(d, box) > 0):
                    d = listreplace(d, box, listentry(d, box) - 1)
                pc = next
            else:
                pc = enext
    return(listentry(d,0))
````

例３のプログラムにおいて，データ d のリストを listcode によって一つの自然数として表現したもの dd は，次のようにして得られる。

````python
d = [0,5,3]
dd = listcode(d)
print(dd)
````

mm : プログラム m の，コード（という数）

また，リスト m に対して，m の要素それぞれに f を適用してできたリストは，
```
list (map (f, m))
```
で得られる。よって，m の各要素に listcode を適用して得られるリストは，次のようにして計算できる。

````python
m = [[2,1,1,2],
 [1,0,0],
 [2,2,3,4],
 [1,0,2],
 [0]
]
m0 = list (map(listcode, m))
print(m0)
````

このリストに対して，もう一度 listcode を適用することにより，
プログラム m のコードとなる数が得られる。

````python
mm = listcode (m0)
print(mm)
````

mm と dd を引数にして univm を呼び出すと，5 + 3 が計算される。

````python
univm(mm, dd)
````
